```python
well_points=potohar_full_points.drop(["OW_ID"],axis=1)     #This will be my final dataframe
```

```python
well_points.head()
```

```python
from scipy.spatial import KDTree
```

```python
min_long, max_long= well_points["LONG"].min(), well_points["LONG"].max()
min_lat, max_lat= well_points["LAT"].min(), well_points["LAT"].max()
```

```python
coords_well=gdf[["LONG","LAT"]].values
```

```python
coords_well
```

```python
tree=KDTree(coords_well)
```

```python
pseudo_zeros=[]
needed=1100

while len(pseudo_zeros) < needed:
  long=rng.uniform(min_long,max_long)
  lat=rng.uniform(min_lat,max_lat)

  dist,_= tree.query([long,lat])
  if dist > 0.02:                  #distance set at 0.02 so as to not be too close and not be too far from the main points.
    pseudo_WL_MBGL=rng.uniform(40,100)
    pseudo_zeros.append([long,lat,pseudo_WL_MBGL])

    #The new points generated have water level depths above 40 so now we are able to create target feature "potential" of 0
```

```python
np.shape(pseudo_zeros)
```

```python
df_zeros=pd.DataFrame(pseudo_zeros,columns=["LONG","LAT","WL_MBGL"])
```

```python
df_zeros.head()
```

```python
pseudo_zeros_geometry=[Point(xy) for xy in zip(df_zeros["LONG"],df_zeros["LAT"])]
```

```python
pseudo_gdf=gpd.GeoDataFrame(df_zeros,geometry=pseudo_zeros_geometry, crs="EPSG:4326")
```

```python
pseudo_gdf.head()
```

```python
well_points.shape
```

```python
pseudo_gdf.shape
```

```python
final_gdf=pd.concat([well_points,pseudo_gdf],ignore_index=True)
```

```python
final_gdf.shape
```

```python
final_gdf.loc[1626]
```

```python
final_gdf["target"]=(final_gdf["WL_MBGL"]<=20).astype(int)
```

```python
final_gdf.drop(["target"],axis=1,inplace=True)
```

```python
final_gdf["potential"]=(final_gdf["WL_MBGL"]<=20).astype(int)          #The limit is set at 20, anything above it starts to become to expensive to drill for small scale organizations and communities.
```

```python
final_gdf["potential"].value_counts()
```
